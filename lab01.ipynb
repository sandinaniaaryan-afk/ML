```python
def find_pairs():
  ls=[2,7,4,1,3,6]
  ls.sort()
  pairs=[]
  i=0
  j=len(ls)-1
  while i<j:
    if ls[i]+ls[j]==10:
      pairs.append((ls[i],ls[j]))
      i+=1
      j-=1
    elif ls[i]+ls[j]<10:
      i+=1
    else:
      j-=1
  return pairs


print(len(find_pairs()))
print(find_pairs())



```

```python
def multiply(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]

    return C


n = int(input("Enter the order of the matrix: "))

print("Enter the matrix:")

A = [list(map(int, input().split())) for _ in range(n)]

m = int(input("Enter the power: "))


result = [[1 if i == j else 0 for j in range(n)] for i in range(n)]


for _ in range(m):
    result = multiply(result, A)

print(f"A^{m} =")
for row in result:
    print(*row)
```

```python
s = input("Enter a string: ")

freq = {}

for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

max_freq = 0

for ch in freq:
    if freq[ch] > max_freq:
        max_freq = freq[ch]

for ch in freq:
    if freq[ch] == max_freq:
        print("Character:", ch)
        print("Frequency:", max_freq)
```
